# Contorl System #
Control System is a Python library for designing and analyzing control systems.
- Types of control systems include:
1. if-else statements
2. ternary operators
3. match-case statements

# if statement #
'''
1. if condition:
-    (# code block)
-    pass
2. else:
-    (# code block)
-    pass
'''

```python
# if Satatement #

a = 10  
b = 20
if a < b:
    print("a is less than b")

# if-else Statement #
a = 10
b = 20
if a < b:
    print("a is less than b")
else:
    print("a is not less than b")

# if-elif-else Statement #
a = 10
b = 20
if a < b:
    print("a is less than b")           
elif a > b:
    print("a is greater than b")    
else:    
    print("a is equal to b")
```

```python
# Nested if Statement #
age = int(input("Enter a your age: "))
if age >= 18:
    print("You are an adult.")
    if age >= 65:
        print("You are a senior citizen.")
    else:
        print("You are not a senior citizen.")  
else:  
    print("You are a minor.")
    if age < 13:
        print("You are a child.")
    else:
        print("You are a teenager.")
```

# Ternary Operator #
The ternary operator is a concise way to write an if-else statement in a single line. 
- It has the following syntax:
> result = value_if_true if condition else value_if_false.

```python
# Ternary Operator #
'''The ternary operator is a concise way to write an if-else statement in a single line. It has the following syntax:
result = value_if_true if condition else value_if_false'''
a = 10
b = 20
result = "a is less than b" if a < b else "a is not less than b"
print(result)
```

# Match-case Statement #
Matched-case Statement is a powerful control flow structure introduced in Python 3.10 that allows you to match patterns against values and execute code based on those patterns. It is similar to switch-case statements found in other programming languages but with more advanced features.
- The basic syntax of a match-case statement is as follows:
- match value:
1.  case pattern1:
        (# code to execute if pattern1 matches)
2.  case pattern2:
        (# code to execute if pattern2 matches)
    ...
3.  case _:
        (# code to execute if no patterns match (optional))

```python
# Match-case Statement #

day = input("Enter a day of the week: ")
match day:
    case "Monday":
        print("It's the start of the week.")
    case "Tuesday":
        print("It's the second day of the week.")
    case "Wednesday":
        print("It's the middle of the week.")
    case "Thursday":
        print("It's the fourth day of the week.")
    case "Friday":
        print("It's the end of the workweek.")
    case "Saturday" | "Sunday": 
        print("It's the weekend!")
    case _:
        print("That's not a valid day of the week.")    
    
```
