# Function # 
Function is a block of code which only runs when it is called. You can pass data, known as parameters, into a function. A function can return data as a result.
In Python, a function is defined using the def keyword, followed by the function name and parentheses (). Inside the parentheses, you can specify parameters if needed. The function body is indented below the function definition.
- syntax:
1. def function_name(parameters):
2.    (# function body)
3.    return value

```python
# Function #

def greet(name):
    return f"Hello, {name}!"
print(greet("Alice"))

input_1st = input("Enter first number: ")
input_2nd = input("Enter second number: ")
def add(a, b):
    return a + b
print("The sum is:", add(int(input_1st), int(input_2nd)))
print(add(5, 10))
```
