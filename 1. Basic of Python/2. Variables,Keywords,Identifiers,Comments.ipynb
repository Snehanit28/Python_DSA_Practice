# Variables, Keywords, Identifiers, Comments #
+ Variables are the containers that stores value.
 
+ Key words are the predefined words. We can not use it for any other work.

+ Identifiers  are the names that we provide to variable/function/classes.

+ Comments are the unexpected statements.

```python
# Variables #

# variables are case sensitive means
name = "Sneha"
Name = "Ajoy" # Name and name are not same variable
 
a = 34
b = 38
sum = a + b
print(a)
print(b)     # output will be 34 and 38
print(sum)   # output will be 72
# we can also assign multiple values to multiple variables in one line
x, y, z = 1, 2, 3
print(x)     # output will be 1
print(y)     # output will be 2 
print(z)     # output will be 3 
#in the above examples variables are name, Name, a, b, sum, x, y and z
```

```python
# Keywords #

# example of keywords
int = 10
float = 3.14  
# int and float are keywords in python which are used to define the data type of the variable
```

```python
# Identifiers #

# example of identifiers
name = "Sneha"
age = 22
# name and age are identifiers which are used to identify the variables
''' differnce between variable and identifier is that variable is a container which holds the value
    and identifier is the name which is used to identify the variable '''
```
