# Operators #
Operators are used to perform operations on variables and values.
+ Python divides the operators in the following groups:
1. Arithmetic operators
2. Assignment operators
3. Comparison operators
4. Logical operators
5. Identity operators
6. Membership operators
7. Bitwise operators

# 1. Arithmetic Operators : #
Used to perform mathematical operations on numeric values
1. ( + ) : addition
2. ( - ) : subtraction 
3. ( * ) : multiplication
4. ( / ) : division
5. ( % ) : modulus (returns the remainder of a division)
6. ( ** ) : exponentiation (raises the first operand to the power of the second operand)
7. ( // ) : floor division (returns the quotient of the division, rounded down to the nearest whole number)

```python
# Arithmetic Operators #
a = 10
b = 5
print(a + b) # => 15,  addition
print(a - b) # => 5,  subtraction
print(a * b) # => 50, multiplication
print(a / b) # => 2.0, division
print(a % b) # => 0, modulus
print(a ** b) # => 100000, exponentiation
print(a // b) # => 2, floor division
```

# 2. Assignment Operators : #
Assignment operators are used to assign values to variables.
+ Types of assignment operators are 
1. = (simple assignment operator)
2. +=  (add and assign) 
3. -=  (subtract and assign)
4. *=  (multiply and assign)
5. /=  (divide and assign)
6. %=  (modulus and assign)
7. **= (exponentiation and assign)
8. //= (floor division and assign)

```python
# Assignment Operators #

x = 100
x += 5     # x + 5 => 105 
print(x) # => 105  
x -= 5     # x - 5 => 100
print(x) # => 100
x *= 5     # x * 5 => 500
print(x) # => 500
x /= 5     # x / 5 => 100.0
print(x) # => 100.0
x **= 5    # x ** 5 => 10000000000.0
print(x) # => 10000000000.0
x //= 5    # x // 5 => 2000000000.0
print(x) # => 2000000000.0
x %= 3     # x % 3 => 2.0
print(x) # => 2.0
x = 123
r = x % 10 
print(r) # 
```

# 3. Comparision operators : #
Comparison operators are used to compare two values and return a boolean value (True or False) 
1. ( == ) : equal to
2. ( != )  : not equal to
3. ( > )   : greater than
4. ( < )   : less than
5. ( >= )  : greater than or equal to
6. ( <= )  : less than or equal to 

```python
# Comparison Operators #

a = 10
b = 5
print(a == b) # => False, equal to
print(a != b) # => True, not equal to
print(a > b) # => True, greater than
print(a < b) # => False, less than
print(a >= b) # => True, greater than or equal to
print(a <= b) # => False, less than or equal to
```

# 4. Logical operators : #
Logical operators are used to combine conditional statements.
+ The logical operators are:
1. and - Returns True if both statements are true  
2. or - Returns True if one of the statements is true
3. not - Reverse the result, returns False if the result is true

```python
# Logical Operators #

a = True
b = False
print(a and b) # => False, logical AND
print(a or b) # => True, logical OR 
print(not a) # => False, logical NOT
print(not b) # => True, logical NOT
```

# 5. Identity Operators : #
Identity operators are used to compare the memory location of two objects.
+ Types of identity operators are (is and is not)

```python
# Identity Operators #

a = 10
b = 10
c = 20
print(a is b) # => True, identity operator returns True if both variables refer to the same object
print(a is c) # => False, identity operator returns False if both variables refer to different objects
print(a is not c) # => True, identity operator returns True if both variables refer to different objects
```

# 6. Membership Operators : #
Membership operators are used to test if a value is present in a sequence (list, tuple, string) or not. 
   + There are two membership operators in Python:
   1.  in: returns True if the value is found in the sequence, otherwise returns False.
   2.  not in: returns True if the value is not found in the sequence, otherwise returns False. 

```python
# Membership Operators #

list1 = [1, 2, 3, 4, 5]

print(3 in list1) # => True, membership operator returns True if value is in the sequence
print(6 not in list1) # => True, membership operator returns True if value is not in the sequence
```

# 7. Bitwise Operators : #
Bitwise operators are used to perform bitwise operations on binary numbers.
+ They operate on the individual bits of the numbers. The common bitwise operators are:
1. & (AND),
2. | (OR),
3. ^ (XOR),
4. ~ (NOT),
5. << (left shift),
6. . >> (right shift) 

```python
# Bitwise Operators #

a = 10 # 1010 in binary
b = 5  # 0101 in binary

print(a & b) # => 0, bitwise AND # 1010 & 0101 => 0000
print(a | b) # => 15, bitwise OR # 1010 | 0101 => 1111
print(a ^ b) # => 15, bitwise XOR # 1010 ^ 0101 => 1111
print(~a) # => -11, bitwise NOT # ~1010 => ...1101
print(a << 1) # => 20, left shift # 1010 << 1 => 10100
print(a >> 1) # => 5, right shift # 1010 >> 1 => 0101
```
