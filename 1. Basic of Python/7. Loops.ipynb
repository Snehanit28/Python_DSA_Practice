# Loops #
Loops are used to repeat a block of code multiple times. In Python,
- There are two main types of loops: for loops and while loops.
1. For Loops
2. While Loops

# For loop #
For loops are used to iterate over a sequence (like a list, tuple, string, etc.) or other iterable objects.
- The syntax of a for loop is:
1. for variable in iterable:
2.   ( # code block to be executed )
Here, 'variable' takes the value of each item in the 'iterable' one by one, and the code block is executed for each item.

```python
# For loop #

for i in range(5):
    print(i, end=' ')
```

```python
list1 = [1, 2, 3, 4, 5]

for i in list1:
    print(i, end=' ')
```

```python
for i in range(1, 11):
    print(i, end=' ')
print()
for i in range(5, 11, 2):
    print(i, end=' ')
print()
for i in range(10, 0, -1):
    print(i, end=' ')
```

# While loop #
While loop is used to execute a block of code repeatedly until a certain condition is met. The syntax of a while loop is as follows:
1. while condition:
2.    (# code block to be executed)
3.    (# The loop continues as long as the condition evaluates to True.)

```python
# While loop #

i = 0
while i < 5:
    print(i, end=' ')
    i += 1
```
