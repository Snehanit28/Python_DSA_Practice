# Input and Outputs #

```python
# Input #
# A single line containing a string S, the name of the person to greet.
name = input("Enter your name: ")
age = int(input("Enter your age: "))
address = 'kolkata'
# Output #
# A single line containing the string "Hello, S!" where S is replaced with the name of the person to greet.
print(f"Hello, {name}!")
# A single line containing the string "You                                                         are A years old." where A is replaced with the age of the person.
print(f"You are {age} years old.")
print(name, age, address, sep=", ")
print(name,end="**")
print(age)
print(address)
```
