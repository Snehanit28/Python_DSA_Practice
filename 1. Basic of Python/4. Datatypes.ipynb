# Data types : #
There are 5 data types in python :
1. Numerical (int,float,complex)
2. String (str)
3. Boolean (bool)
4. None 
5. Data structures (list,tuple,dictionaries,sets etc.)

# 1. Numerical (int,float,complex) #
Numerical data types are used to store numeric values. There are three types of numerical data types in python - int, float and complex. 
 1. int : int data type is used to store integer values. It can be positive, negative or zero. It does not have any decimal point. 
 2. float : float data type is used to store decimal values. It can be positive, negative or zero. It has a decimal point.
 3. complex : complex data type is used to store complex numbers. It has a real part and an imaginary part. The real part is represented by a float and the imaginary part is represented by a float followed by the letter 'j'. 

```python
# Numerical (int,float,complex) #

a = 10 # int
b = 3.14 # float
c = 2 + 3j # complex
print(a,type(a)) # => 10 <class 'int'>
print(b,type(b)) # => 3.14 <class 'float'>
print(c,type(c)) # => (2+3j) <class 'complex'>
```

# 2. String (str) #
String is a sequence of characters enclosed in single quotes, double quotes or triple quotes.

```python
# String (str) #

Grade = 'A'
print(Grade,type(Grade)) # => A <class 'str'>

name = "Sneha"
print(name,type(name)) # => Sneha <class 'str'>

course = 'Python Programming'
print(course,type(course)) # => Python Programming <class 'str'>
```

# 3. Boolean (bool) #
Boolean data type can have only two values: True or False. It is used to represent the truth value of an expression.

```python
#  Boolean (bool) #

is_student = True
print(is_student,type(is_student)) # => True <class 'bool'>
```

# 4. None #
None is a special data type in Python that represents the absence of a value or a null value. It is often used to indicate that a variable has no value or to signify the end of a list or other data structure. The NoneType is the type of the None object.

```python
# None (NoneType) #

value = None
print(value,type(value)) # => None <class 'NoneType'>
```

# 5. Data structures (list,tuple,dictionaries,sets etc.) #
 Data structures are used to store and organize data in a way that allows for efficient access and modification.
1. List : A list is an ordered collection of items that can be of different types. Lists are mutable, meaning you can change their content after they are created. 
2. Tuple : A tuple is an ordered collection of items that can be of different types. Tuples are immutable, meaning you cannot change their content after they are created.
3. Dictionary : A dictionary is an unordered collection of key-value pairs. Dictionaries are mutable, meaning you can change their content after they are created.
4. Set : A set is an unordered collection of unique items. Sets are mutable, meaning you can change their content after they are created. 

```python
#  Data structures (list,tuple,dictionaries,sets etc.) #

list1 = [1, 2, 3, 4, 5] # list
tuple1 = (1, 2, 3, 4, 5) # tuple
dict1 = {
    "name": "Sneha",
    "age": 25,
    "course": "Python Programming"
    } # dictionary   
set1 = {1, 2, 3, 4, 5} # set
print(list1,type(list1)) # => [1, 2, 3, 4, 5] <class 'list'>
print(tuple1,type(tuple1)) # => (1, 2, 3, 4, 5) <class 'tuple'>
print(dict1,type(dict1)) # => {'name': 'Sneha', 'age': 25, 'course': 'Python Programming'} <class 'dict'>
print(set1,type(set1)) # => {1, 2, 3, 4, 5} <class 'set'>
```
