# Recursion #
The process in which a function calls itself directly or indirectly is called recursion and the corresponding function is called Recursive function.
1. Base case
2. Recursive case 

```python
# Print 1 to n using Recursion 

def printNumbers(i,n):
    # base case-
    if i>n:
        return
    # recursive case-
    print(i,end=" ")
    printNumbers(i+1,n)

printNumbers(1,5)
```

```python
# factorial

def fact(n):
    if n==0:
        return 1
    return n*fact(n-1)
print(fact(4))
```

```python
# Recursive stack

def fun(n):
    if n==0:
        return
    fun(n-1)
    print(n,end=" ")
fun(4)
```

```python
def gcd(a,b):
    if b==0:
        return a
    return gcd(b,a%b)
def lcm(a,b):
    return a*b/gcd(a,b)

print(gcd(120,10))
print(lcm(15,50))
```
