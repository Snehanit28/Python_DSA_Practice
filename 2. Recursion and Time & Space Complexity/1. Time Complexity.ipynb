# Time Complexity #
Time complexity in computer science describeshow the excution time of the algorithm grows as the input size increases. It essentially quantifies the number of operations an alogorith performs as a function of the input size data.
## Asymptotic notations ##
1. Best case (Ω)
2. Average case (θ)
3. Worst case (O)
## Types of time complexity ##
Common types include 
1. Constant time (O(1)),
2. Logarithmic time (O(log n)),
3. Linear time (O(n)),
4. Quadratic time (O(n^2)),
5. Exponential time (O(2^n)).
6. others - ( Linearithmic time(O(n log n)), Cubic time(O(n^3)), Factorial time(O(n!)) )

> O(1) < O(log n) < O(n) > O(nlog n) < O(n^2) < O(n^3) < O(2^n) < O(n!)

```python
# O(1) # Constant time complexity

for i in range(100):
    print(i,end=" ")
```

```python
# O(log n) # Logarithmic time complexity

n = 100

while n>0:
    print(n,end=" ")
    n//=2
print( )

i = 1
m = 100
while i<=m:
    print(i,end=" ")
    i*=2
```

```python
# O(n) # Linear time complexity

n = int(input("Enter the number: "))
print(f"n = {n}")
m = int(input("Enter a number: "))
print(f"m = {m}")

for i in range(n):
    print(i,end=" ")
print( )
for i in range(n+m):
    print(i,end=" ")
print( )
for i in range(n-m):
    print(i,end=" ")
print( )
for i in range(n*m):
    print(i,end=" ")
print( )
for i in range(int(n/m)):
    print(i,end=" ")

# Time Complexity -
# O(n+n+n+n+n) = O(5n)
# O(5n) ~ O(n)
```

```python
# O(n^2) # Quadratic time complexity

n = int(input("Enter the number: "))
print(f"n = {n}")

for i in range(n**2):
    print(i,end=" ")   # O(n^2)
print( )
for i in range(n):
    for j in range(n):
        print(i,j)     # O(n^2)
for i in range(m):
    print(i)      # O(n)

# Time complexity
# O((n^2)+(n^2)+(n)) = O(2(n^2)+(n))
# O(2(n^2)+(n)) ~ O(n^2)
```
