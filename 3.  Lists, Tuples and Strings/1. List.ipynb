### *1. List in Python* ###
In python, list allows us to store multiple items in a single variable.
- Lists are similar to arrays(dynamic arrays that allow us to store items of different data types) in other programming languages. Lists are hetrogenious.(We can store intiger,float and string in one list)
- We use "[ ]" to represent list in python.

##### *List Characteristic* #####
In python lists are:

- Ordered - They maintain the order of elements.
- Mutable - Items can be changed after creation.
- Allow duplicates - They can contain duplicate values.


```python
student=['Sneha',20,'A',153.67] # Ordered- maintain the order of elemnts and it print the list as it is.
print(student)
print(type(student))
names=['sneha','souvik','sonia','rahul']
print(names)
print(type(names))
print(names[-1]) # negative indexing
print(names[0])
names[0]= 'Ayan' # Mutable- items can be changed after creation.
names
num=[1,3,2,1,4,2,3,5] # Allow duplicates - They can contain duplicate values.
print(num)
```

##### Operations #####
- len()
- negative indexing
- append()
- extend()
- insert()
- remove()
- pop()

```python
list1 = [23,'hello',42.134,True,214]
print(len(list1))      # length
print(list1[2])        # indexing
print(list1[-4])       # negative indexing
print(type(list1)) 
list1
```

```python
num = [1,6,3,7,4,8]
num.append(5)          # append()
num.append([1,2,3])
num
```

```python
num.extend([11,10,13])  # extend()
print(num)        
print(num[7])
print(num[7][2])
```

```python
num.insert(3,9)         # insert()
print(num)
```

```python
num.remove(7)           # remove() # remove element
num
```

```python
num.pop(2)              # pop()  # remove index
num
```

```python
num.clear()             # clear() clear all element
num
```

##### Function #####
- min()
- max()
- slicing
- count()
- sort()
- reverse()
- copy()
- index()

```python
list2 = [4,7,3,6,8,2,9,1,0,5,10]
```

```python
max(list2)   # o(n)
```

```python
min(list2)   # o(n)
```

```python
list2.reverse()  # reverse()  # reverse the list  # O(n)
list2
```

```python
list2.sort() # sort()  # sort the list in ascending order # O(n logn)   ``
list2
```

```python
list3 =[1,4,3,1,5,2,4,2,1,5,3,6,1,4]
```

```python
list3.index(4)    # index()  # find the index of first occurrence of 4 
list3.index(4,2)  # index()  # find 4 after index 2
```

```python
list3.count(1)    # 4
list3.count(4)    # 3
```

```python
list1 = [1,2,3]
list2 = list1  # deep copy
list2.append(4)
print(list1,id(list1))
print(list2,id(list2))
```

```python
list1 = [1,2,3]
list2 = list1.copy()  # shallow copy
list2.append(4)
print(list1,id(list1))
print(list2,id(list2))
```

| Feature            | Subarray (Continuous)                          | Subsequence (Not Continuous)                          |
|-------------------|-----------------------------------------------|------------------------------------------------------|
| Definition        | A contiguous part of the array                | Elements taken in order but not necessarily continuous |
| Continuity        | Continuous only                              | Can skip elements                                    |
| Formation         | Select continuous segment                    | Delete zero or more elements                         |
| Total Possible    | n(n+1)/2                                     | 2^n                                                  |
| Example 1         | [1, 3, 2]                                     | [6, 2, 3]                                            |
| Example 2         | [3, 2]                                        | [5, 6]                                               |
| Example 3         | [2, 6, 5, 8]                                  | [8, 5, 6]                                            |
| Invalid Example   | [1, 2, 5] (not continuous)                   | [3, 1] (wrong order)                                 |


```python
# Slicing
list1 = [4,7,3,6,8,2,9,1,0,5,10]
print(list1[2:6])
print(list1[2:8:2])
```

```python
print(list1[::-1])
print(list1[4::-2])
print(list1[4::2])
print(list1[:8:3])
```
